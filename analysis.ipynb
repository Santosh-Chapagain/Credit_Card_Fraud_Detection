```python
import os 
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 

from sklearn.preprocessing import StandardScaler 
```

### Analysis and preprocessing

```python
data = pd.read_csv("creditcard.csv")
```

```python
data.head()
```

```python
data.shape
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
sc= StandardScaler()
data['Amount']= sc.fit_transform(pd.DataFrame(data['Amount']))
```

```python
data.head()
```

```python
data= data.drop(['Time'], axis=1)
```

```python
data.head()
```

```python
data.duplicated().any()
```

```python
data = data.drop_duplicates()
```

### Handling Imbalenced data

```python
data['Class'].value_counts()
```

```python
import matplotlib.pyplot as plt

counts = data["Class"].value_counts()

plt.bar(counts.index.astype(str), counts.values)
plt.xlabel("Class")
plt.ylabel("Count")
plt.title("Class Distribution")
plt.show()
```

```python
x = data.drop('Class', axis=1)
y= data['Class']
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test= train_test_split(x, y, test_size= 0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
lr= LogisticRegression()
lr.fit(x_train, y_train)
y_pred= lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
print("Accuracy Score: ", accuracy_score(y_test, y_pred))
print("Precision_Score: ", precision_score(y_test, y_pred))
print("Recall_Score: ", recall_score(y_test, y_pred))
print("F1_Score: ", f1_score(y_test, y_pred))
```

### Handling Imbalanced dataset

```python
#undersampling
#oversampling 
```

### Undersampling

```python
fraud_data =  data[data["Class"] == 1]
```

```python
fraud_data['Class'].value_counts()
```

```python
normal_data= data.sample(473)
```

```python
normal_data['Class'].value_counts()
```

```python
undersample_data= pd.concat([fraud_data, normal_data], ignore_index= True)
```

```python
undersample_data.head()
```

```python
undersample_data['Class'].value_counts()
```

```python
u_x = undersample_data.drop('Class', axis=1)
u_y = undersample_data['Class']
```

```python
x_train2, x_test2, y_train2, y_test2= train_test_split(u_x, u_y, test_size= 0.2, random_state= 42)
```

### Logistic Regression

```python
lr2 = LogisticRegression()
lr2.fit(x_train2, y_train2)
y_pred2 = lr2.predict(x_test2)
```

```python
print("Accuracy Score: ", accuracy_score(y_test2, y_pred2))
print("Precision_Score: ", precision_score(y_test2, y_pred2))
print("Recall_Score: ", recall_score(y_test2, y_pred2))
print("F1_Score: ", f1_score(y_test2, y_pred2))
```

### Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
dt= DecisionTreeClassifier(min_samples_leaf=4)
dt.fit(x_train2, y_train2)
y_pred2= dt.predict(x_test2)
```

```python
print("Accuracy Score: ", accuracy_score(y_test2, y_pred2))
print("Precision_Score: ", precision_score(y_test2, y_pred2))
print("Recall_Score: ", recall_score(y_test2, y_pred2))
print("F1_Score: ", f1_score(y_test2, y_pred2))
```

### Randomforest Classifier

```python
from sklearn.ensemble import RandomForestClassifier
rf= RandomForestClassifier()
rf.fit(x_train2, y_train2)
y_pred2= rf.predict(x_test2)
```

```python
print("Accuracy Score: ", accuracy_score(y_test2, y_pred2))
print("Precision_Score: ", precision_score(y_test2, y_pred2))
print("Recall_Score: ", recall_score(y_test2, y_pred2))
print("F1_Score: ", f1_score(y_test2, y_pred2))
```

### Oversampling

```python
data.head()
```

```python
data.shape
```

```python
x= data.drop('Class', axis= 1)
y= data['Class']
```

```python
from imblearn.over_sampling import SMOTE
```

```python
x_res, y_res= SMOTE().fit_resample(x,y)
```

```python
y_res.value_counts()
```

```python
x_res.head()
```

```python
x_train3, x_test3, y_train3, y_test3= train_test_split(x_res, y_res, test_size= 0.2, random_state=42)
```

```python
x_train3.shape
```

### Logistic regression

```python
lr3= LogisticRegression()
lr3.fit(x_train3, y_train3)
y_pred3= lr3.predict(x_test3)
```

```python
print("Accuracy Score: ", accuracy_score(y_test3, y_pred3))
print("Precision_Score: ", precision_score(y_test3, y_pred3))
print("Recall_Score: ", recall_score(y_test3, y_pred3))
print("F1_Score: ", f1_score(y_test3, y_pred3))
```

### Decision-Tree Classifier

```python
dt3= DecisionTreeClassifier()
dt3.fit(x_train3,y_train3)
y_pred3= dt3.predict(x_test3)
```

```python
print("Accuracy Score: ", accuracy_score(y_test3, y_pred3))
print("Precision_Score: ", precision_score(y_test3, y_pred3))
print("Recall_Score: ", recall_score(y_test3, y_pred3))
print("F1_Score: ", f1_score(y_test3, y_pred3))
```

### Random Forest Classifier

```python
rf3= RandomForestClassifier()
rf3.fit(x_train3, y_train3)
y_pred3= rf3.predict(x_test3)
```

```python
print("Accuracy Score: ", accuracy_score(y_test3, y_pred3))
print("Precision_Score: ", precision_score(y_test3, y_pred3))
print("Recall_Score: ", recall_score(y_test3, y_pred3))
print("F1_Score: ", f1_score(y_test3, y_pred3))
```
